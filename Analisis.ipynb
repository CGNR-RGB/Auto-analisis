```python
import pandas as pd
```

```python
# URL del archivo Excel de INEGI
url = "https://www.inegi.org.mx/contenidos/datosprimarios/iavl/tabulados/8_Ventas_serie.xlsx"

# Descargar y leer el Excel
autos = pd.read_excel(url)
```

```python
# Consultar los primeros registros
autos.head(1)
```

```python
# Revisar dimensiones y columnas
print(autos.shape)
print(autos.columns)
```

```python
## Extraer la fila cuatro y todas las columnas porque ahi se encuentran los nombres
nombres = autos.iloc[4,:]
nombres
```

```python
# Seleccionar desde la fila 5 y todas las columnas 
datos = autos.iloc[5:,:]
```

```python
datos.head(1)
```

```python
datos.columns = nombres
```

```python
datos.head(1)
```

```python
# La columna Tema solo tiene basura
datos['Tema'].unique()
```

```python
## Eliminar la columna tema

datos.drop(['Tema'], axis=1, inplace=True)
```

```python
datos.head(1)
```

```python
### Además tengo más cambios 
```

```python
datos['Mes'].unique()
```

```python
## Mapear los datos 
meses = {'Enero':1,'Febrero':2,'Marzo':3,'Abril':4,'Mayo':5,
         'Junio':6,'Julio':7,'Agosto':8,'Septiembre':9,'Octubre':10,'Noviembre':11,
         'Diciembre':12}
meses
```

```python
columnas =  datos.columns
columnas
```

# Revisar el número de valores faltantes

```python
## Faltantes por columnas 
for columna in columnas:
    print(sum(datos[columna].isna()))
```

### Revisar el tipo de datos 

```python
datos.dtypes
```

```python
## Ver las fials con valores faltantes 

datos[datos['Año'].isna()]
```

```python
datos = datos.dropna()
```

```python
## Faltantes por columnas 
for columna in columnas:
    print(sum(datos[columna].isna()))
```

```python
datos['Mes'] = datos['Mes'].map(meses)
```

```python
## Crear el campo de fecha 
datos["date"] = pd.to_datetime(datos["Año"].astype('str')+ "-" + datos["Mes"].astype('str')+'-01', format="%Y-%m-%d")
```
